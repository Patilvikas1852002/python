### 🐼 Lesson 2: Creating DataFrames (Multiple Ways)
### ✅ Method 1: From Dictionary (MOST COMMON)

```python
import pandas as pd

data = {
    "Name": ["Amit", "Sara", "Raj"],
    "Age": [25, 30, 22],
    "City": ["Pune", "Mumbai", "Delhi"]
}

df = pd.DataFrame(data)
print(df)
```

### ✅ Method 2: From List of Lists

```python

data = [
    ["Amit", 25, "Pune"],
    ["Sara", 30, "Mumbai"],
    ["Raj", 22, "Delhi"]
]

df = pd.DataFrame(data, columns=["Name", "Age", "City"])
df
```

### ✅ Method 3: From List of Dictionaries

```python
data = [
    {"Name": "Amit", "Age": 25},
    {"Name": "Sara", "Age": 30}
]

df = pd.DataFrame(data)
df
```

### ✅ Method 4: From NumPy Array

```python
import numpy as np

arr = np.array([[1,2], [3,4]])
df = pd.DataFrame(arr, columns=["A", "B"])
```
