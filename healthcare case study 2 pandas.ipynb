#### 📊 Case Study: Hospital Patient Records Analysis

```python
import pandas as pd

data = {
    "PatientID": [101,102,103,104,105,106,107,108,109,110],
    "PatientName": ["Rahul","Anita","Vikram","Sneha","Arjun","Pooja","Karan","Meera","Ravi","Neha"],
    "Age": [45,32,60,28,50,41,36,55,47,30],
    "Gender": ["M","F","M","F","M","F","M","F","M","F"],
    "City": ["Mumbai","Delhi","Mumbai","Pune","Delhi","Mumbai","Pune","Mumbai","Delhi","Pune"],
    "Disease": ["Diabetes","Flu","Hypertension","Flu","Diabetes","Asthma","Flu","Hypertension","Asthma","Flu"],
    "Doctor": ["Dr. Shah","Dr. Mehta","Dr. Shah","Dr. Rao","Dr. Mehta","Dr. Shah","Dr. Rao","Dr. Shah","Dr. Mehta","Dr. Rao"],
    "TreatmentCost": [5000,800,7000,900,5500,3000,850,7200,3200,950],
    "VisitDate": pd.to_datetime([
        "2024-01-02","2024-01-05","2024-01-08","2024-01-10","2024-01-15",
        "2024-02-01","2024-02-05","2024-02-10","2024-02-12","2024-02-15"
    ]),
    "DaysAdmitted": [2,0,3,0,2,1,0,4,1,0]
}

df = pd.DataFrame(data)
df
```

#### 🧮 PART 3 — Feature Engineering

```python
#Q11. Create column CostPerDay

#(Hint: TreatmentCost / DaysAdmitted; handle zero days)
df["CostPerDay"] = df["TreatmentCost"] / df["DaysAdmitted"].replace(0, pd.NA)
```

```python
#Q12. Add VisitMonth column
df["VisitMonth"]=df["VisitDate"].dt.month
df
```

```python
#Q13. Create AgeGroup column:

#< 30 → Young

#30–50 → Adult

#50 → Senior
df["AgeGroup"] = df["Age"].apply(
    lambda x: "Young" if x < 30 else "Adult" if x <= 50 else "Senior"
)
df
```

#### 📊 PART 4 — Aggregation & Analysis

```python
#Q14. Total treatment cost
df["TreatmentCost"].sum()
```

```python
#Q15. Total cost by city
df.groupby("City")["TreatmentCost"].sum()
```

```python
#Q16. Total cost by disease
df.groupby("Disease")["TreatmentCost"].sum()
```

```python
#Q17. Average treatment cost by doctor

df.groupby("Doctor")["TreatmentCost"].mean()  #sum()
```

```python
#Q18. Total admitted days per disease
df.groupby("Disease")["DaysAdmitted"].sum()
```
