```python
import pandas as pd
import numpy as np

np.random.seed(42)

# possible values
products = ["Laptop", "Phone", "Headphones", "Shoes", "Watch",
            "Backpack", "T-shirt", "Keyboard", "Mouse", "Tablet"]

categories = {
    "Laptop": "Electronics",
    "Phone": "Electronics",
    "Headphones": "Electronics",
    "Tablet": "Electronics",
    "Keyboard": "Electronics",
    "Mouse": "Electronics",
    "Shoes": "Fashion",
    "Watch": "Fashion",
    "Backpack": "Accessories",
    "T-shirt": "Fashion"
}

cities = ["Pune", "Mumbai", "Delhi", "Bangalore", "Hyderabad", "Chennai"]
payment_types = ["UPI", "Credit Card", "Debit Card", "Cash on Delivery"]

n = 120

data = {
    "order_id": np.arange(1001, 1001+n),
    "customer_id": np.random.randint(200, 260, n),
    "order_date": pd.to_datetime("2025-01-01") + pd.to_timedelta(np.random.randint(0, 90, n), unit="D"),
    "product": np.random.choice(products, n),
    "price": np.random.randint(500, 60000, n),
    "quantity": np.random.randint(1, 5, n),
    "city": np.random.choice(cities, n),
    "payment_type": np.random.choice(payment_types, n),
    "returned": np.random.choice([True, False], n, p=[0.1, 0.9])
}

df = pd.DataFrame(data)

# map categories
df["category"] = df["product"].map(categories)

# reorder columns
df = df[[
    "order_id", "customer_id", "order_date", "product",
    "category", "price", "quantity", "city",
    "payment_type", "returned"
]]

df.head()
```

```python
#find total revenue generated.
df["TotalRevenue"]=df["price"] * df["quantity"].sum()
df.head()
```

```python
#dentify the top 5 selling products by quantity.

df.groupby("product")["quantity"].sum().sort_values(ascending=False).head(5)
```

```python
top_products = (
    df.groupby("product", as_index=False)["quantity"]
      .sum()
      .sort_values(by="quantity", ascending=False)
      .head(5)
)

print(top_products)
```

```python
#Which category generates the highest revenue?
df.groupby("category")["TotalRevenue"].sum().sort_values(ascending=True)
```

```python
#Find customers who made more than 5 purchases.
#df.groupby("customer_id")["order_id"].count().sort_values(ascending=False).head(5)    
#✔ shows top 5 buyers
#❌ may include customers with ≤ 5 orders
#❌ misses other customers with >5 orders

order_counts = df.groupby("customer_id")["order_id"].count()

customers_gt5 = order_counts[order_counts > 5]

print(customers_gt5)
```

```python
#Calculate return rate per category.
#df.groupby("category")["returned"].mean()
#
df.groupby("category")["returned"].mean() * 100
```

```python
#Which city has the highest average order value?

df["order_value"] = df["price"] * df["quantity"]
df.groupby("city")["order_value"].mean().sort_values(ascending=False)
```

```python
#Detect duplicate orders.
#✅ 1️⃣ Check if duplicates exist
df.duplicated().sum()
#✅ 2️⃣ View duplicate rows
#df[df.duplicated()]
#✅ 4️⃣ Check duplicates based on order_id (most common)
df[df.duplicated(subset="order_id", keep=False)]
#✅ 5️⃣ Count duplicate order IDs
df["order_id"].duplicated().sum()
#✅ 6️⃣ Remove duplicates
df = df.drop_duplicates()
```

```python
#Find orders placed on weekends.
df[df["order_date"].dt.day_name().isin(["Saturday", "Sunday"])]
```

```python
#Create a pivot table showing revenue by city vs category.
pd.pivot_table(
    df,
    values="order_value",
    index="city",
    columns="category",
    aggfunc="sum",
    fill_value=0
)
```
