```python
import pandas as pd

data = {
    "OrderID": range(1001, 1021),
    "OrderDate": [
        "2024-01-01","2024-01-02","2024-01-03","2024-01-04","2024-01-05",
        "2024-01-06","2024-01-07","2024-01-08","2024-01-09","2024-01-10",
        "2024-01-11","2024-01-12","2024-01-13","2024-01-14","2024-01-15",
        "2024-01-16","2024-01-17","2024-01-18","2024-01-19","2024-01-20"
    ],
    "Customer": ["Amit","Neha","Rahul","Sneha","Vikram","Amit","Neha","Rohit","Pooja","Amit",
                 "Riya","Rahul","Sneha","Vikram","Amit","Rohit","Pooja","Neha","Riya","Rahul"],
    "City": ["Pune","Mumbai","Delhi","Pune","Bangalore","Pune","Mumbai","Delhi","Pune","Pune",
             "Mumbai","Delhi","Pune","Bangalore","Pune","Delhi","Pune","Mumbai","Mumbai","Delhi"],
    "Category": ["Electronics","Clothing","Electronics","Groceries","Clothing","Electronics",
                 "Groceries","Clothing","Electronics","Groceries","Clothing","Electronics",
                 "Groceries","Clothing","Electronics","Groceries","Electronics","Clothing",
                 "Electronics","Groceries"],
    "Product": ["Laptop","T-Shirt","Mobile","Rice Bag","Jeans","Headphones","Oil","Shirt",
                "Tablet","Milk","Jacket","Camera","Wheat","Trousers","Speaker","Sugar",
                "Smartwatch","Dress","Powerbank","Tea"],
    "Quantity": [1,3,1,5,2,2,3,2,1,4,1,1,6,2,1,3,1,2,2,5],
    "Price": [55000,800,20000,1200,1500,2500,900,1000,15000,600,2200,30000,1800,1700,4000,700,8000,1400,1200,500],
    "Discount": [5000,0,2000,100,0,200,50,0,1500,0,100,3000,200,0,500,0,1000,100,0,50],
    "Rating": [5,4,5,3,4,4,3,4,5,3,4,5,3,4,5,3,5,4,4,3],
    "PaymentMethod": ["UPI","COD","Card","UPI","Card","UPI","COD","Card","UPI","UPI",
                      "Card","Card","COD","UPI","UPI","COD","Card","UPI","Card","COD"]
}

df = pd.DataFrame(data)
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

df.head()
```

#### 🎯 SECTION 1 — Basic Exploration

```python
### Q1. Total number of orders
len(df)
#df.shape[0]
```

```python
### Q2. Total revenue (before discount)
#(df["Price"]* df["Quantity"]).sum()
(df["Price"] * df["Quantity"]).sum()
```

```python
### Q3. Total discount given
(df["Discount"] * df["Quantity"]).sum()
```

```python
### Q4. Net revenue after discount
((df["Price"] - df["Discount"]) * df["Quantity"]).sum()
```

```python
### Q5. Average order value
net_revenue = ((df["Price"] - df["Discount"]) * df["Quantity"]).sum()
avg_order_value = net_revenue / len(df)
avg_order_value
```

#### 🎯 SECTION 2 — Customer Insights


```python
df.head(1)
```

```python
### Q6. Top 5 customers by spending

df["NetAmount"] = (df["Price"] - df["Discount"]) * df["Quantity"]

df.groupby("Customer")["NetAmount"].sum() \
  .sort_values(ascending=False) \
  .head(5)
```

```python
### Q7. Customer who placed most orders
# Method 1 
df["Customer"].value_counts().idxmax()

# Customer with most orders
df["Customer"].value_counts().idxmax()
# method using groupby
df.groupby("Customer").size().sort_values(ascending=False).head(1)
```

```python
### Q8. Average rating per customer
df.groupby("Customer")["Rating"].mean()

```

```python
### Q9. Customer lifetime value (CLV)
df["NetAmount"] = (df["Price"] - df["Discount"]) * df["Quantity"]

df.groupby("Customer")["NetAmount"].sum().sort_values(ascending=False)
```

#### 🎯 SECTION 4 — Product & Category Insights

```python
### Q15. Category generating highest revenue
df.groupby("Category")["NetAmount"].sum().sort_values(ascending=False)
```

#### 🎯 SECTION 5 — Time Analysis

```python
### Q19. Orders per day
df["Date"] = pd.to_datetime(df["OrderDate"])

df.groupby(df["Date"].dt.date)["OrderID"].count()
```

```python
### Q20. Day with highest sales
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

df.groupby(df["OrderDate"].dt.date)["NetAmount"].sum().idxmax()
```

```python
### Q21. Weekend vs weekday revenue
df["DayType"] = df["OrderDate"].dt.dayofweek.apply(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)
df.groupby("DayType")["NetAmount"].sum()
```

```python
### Q22. Rolling 3-day average sales
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
daily_sales = df.groupby(df["OrderDate"].dt.date)["NetAmount"].sum()
rolling_avg = daily_sales.rolling(window=3).mean()

rolling_avg
```

#### 🎯 SECTION 6 — Discount Analysis

```python
### Q23. Orders with discount vs without
df["DiscountFlag"] = df["Discount"].apply(
    lambda x: "With Discount" if x > 0 else "No Discount"
)

df["DiscountFlag"].value_counts()
```

```python
### Q24. Does discount increase quantity ordered?
df["DiscountFlag"] = df["Discount"].apply(
    lambda x: "With Discount" if x > 0 else "No Discount"
)
df.groupby("DiscountFlag")["Quantity"].mean()
```

```python
### Q25. Avg discount per category
df.groupby("Category")["Discount"].mean()
```

```python
### Q26. Discount impact on ratings
df.groupby(df["Discount"] > 0)["Rating"].mean()
```

#### 🎯 SECTION 7 — Advanced Pandas Skills

```python
### Q27. Create column: `NetPrice = Price - Discount`

df["NetPrice"] = df["Price"] - df["Discount"]
```

```python
### Q28. Create column: `TotalValue = Quantity * NetPrice`
df["TotalValue"] = df["Quantity"] * df["NetPrice"]
df
```

```python
### Q29. Find top 3 highest value orders
top_orders = df.sort_values(by="TotalValue", ascending=False).head(3)
print(top_orders)
```
