#### 🧮 PART 3 — Feature Engineering

```python
### Q10. Create FinalAmount column
#👉 FinalAmount = OrderValue + DeliveryFee − Discount
df["finalAmount"]=df["OrderValue"] + df["DeliveryFee"] - df["Discount"]
df

```

```python
### Q11. Extract Hour from OrderTime
df["year"] = df["OrderTime"].dt.year
df["month"] = df["OrderTime"].dt.month
df["day"] = df["OrderTime"].dt.day
df["hour"] = df["OrderTime"].dt.hour
df["minute"] = df["OrderTime"].dt.minute
df["second"] = df["OrderTime"].dt.second
```

```python
### Q12. Create TimeSlot:

#* 12–16 → Lunch
#* 17–22 → Dinner
#* otherwise → Other
import numpy as np

conditions = [
    df["OrderTime"].dt.hour.between(12, 16),
    df["OrderTime"].dt.hour.between(17, 22)
]

choices = ["Lunch", "Dinner"]

df["TimeSlot"] = np.select(conditions, choices, default="Other")
```

####  📉 PART 8 — Strategic Insights (Real Analyst Thinking)

```python
### Q29. Which city is most profitable?
df.groupby("City")["OrderValue"].sum().sort_values(ascending=False)
```

```python
### Q30. Which category should be promoted?
df.groupby("Category").agg(
    Orders=("Category", "count"),
    Revenue=("OrderValue", "sum"),
    Avg_Order_Value=("OrderValue", "mean"),
    Avg_Rating=("Rating", "mean"),
    Avg_Discount=("Discount", "mean")
).sort_values(by="Orders", ascending=False)
```

```python
### Q31. Should discounts be increased or reduced?
df.groupby(df["Discount"] > 0).agg(
    Orders=("OrderValue","count"),
    Revenue=("OrderValue","sum"),
    Avg_Order_Value=("OrderValue","mean"),
    Avg_Rating=("Rating","mean")
).rename(index={False:"No Discount", True:"Discounted"})
```

```python
### Q32. When should delivery staffing be increased?
df["hour"] = df["OrderTime"].dt.hour
df.groupby("hour").size().sort_values(ascending=False)
```
