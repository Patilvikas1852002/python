```python
#1 Print a pyramid star pattern (center aligned).
rows = 5

for i in range(1, rows + 1):

    # Print spaces
    for j in range(rows - i):
        print(" ", end="")

    # Print stars
    for k in range(2 * i - 1):
        print("*", end="")

    print()   # Move to next line
```

```python
#2 Print Pascal’s Triangle for n rows.
n = 5

for i in range(n):
    number = 1
    for j in range(n - i - 1):
        print(" ", end="")

    for j in range(i + 1):
        print(number, end=" ")
        number = number * (i - j) // (j + 1)

    print()
```

```python
#3 Find all Armstrong numbers between 1 and 1000.
for num in range(1, 1001):
    order = len(str(num))
    sum = 0
    temp = num

    while temp > 0:
        digit = temp % 10
        sum += digit ** order
        temp //= 10

    if sum == num:
        print(num)
```

```python
#4 Create a 3x3 matrix and print its transpose using loops.
# Creating a 3x3 matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Original Matrix:")
for i in range(3):
    for j in range(3):
        print(matrix[i][j], end=" ")
    print()

print("\nTranspose Matrix:")
for i in range(3):
    for j in range(3):
        print(matrix[j][i], end=" ")
    print()
```

```python
#5 Multiply two matrices using nested for loops. 
# First matrix
A = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Second matrix
B = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]

# Result matrix initialized with zeros
result = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0]
]

# Matrix multiplication using nested loops
for i in range(3):
    for j in range(3):
        for k in range(3):
            result[i][j] += A[i][k] * B[k][j]

# Print result
print("Result Matrix:")
for i in range(3):
    for j in range(3):
        print(result[i][j], end=" ")
    print()
```

```python
#6 Implement Bubble Sort using for loops.
a = [5, 3, 8, 4, 2]

n = len(a)

for i in range(n):
    for j in range(0, n - i - 1):
        if a[j] > a[j + 1]:
            # Swap
            temp = a[j]
            a[j] = a[j + 1]
            a[j + 1] = temp

print("Sorted list:", a)
```

```python
#7 Count frequency of each character in a string using a dictionary and loop.
string = "hello world"

freq = {}

for char in string:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print("Character Frequency:")
for key in freq:
    print(key, ":", freq[key])
```
