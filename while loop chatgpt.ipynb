## 🔹 Basic While Loop Practice

Print numbers from 1 to 10 using a while loop.

Print numbers from 10 down to 1.

Print all even numbers from 1 to 20.

Print all odd numbers from 1 to 20.

Print the square of numbers from 1 to 10.

```python
#Print numbers from 1 to 10 using a while loop.
num=1
while num<=10:
    print(num)
    num+=1
```

```python
#Print numbers from 10 down to 1.
num=10
while num>=1:
    print(num)
    num-=1
```

```python
#Print all even numbers from 1 to 20.
num = 1
while num <= 20:
    if num % 2 == 0:
        print(num)
    num += 1

```

```python
#Print all odd numbers from 1 to 20.
num=1
while num<=20:
    if num%2!=0:
        print(num)
    num +=1
```

```python
#Print the square of numbers from 1 to 10.
num=1
while num<=10:
    square=num*num
    print(square)
    num +=1
```

## 🔹 Sum & Calculations

Find the sum of numbers from 1 to 100.

Find the sum of even numbers from 1 to 50.

Print the multiplication table of 7.

Print the multiplication table of a number stored in a variable.

Find the factorial of a number using a while loop.


```python
#Find the sum of numbers from 1 to 100
num=1
count=0
while num<=100:
    count=num+count
    num=num+1
print(count)
```

```python
#Find the sum of even numbers from 1 to 50.

num = 1
count = 0

while num <= 50:
    if num % 2 == 0:
        count += num
    num += 1

print(count)

num = 1
count=0
while num<=50:
    if num%2==0:
        count+=num
    num +=1
print(count)
```

```python
#Print the multiplication table of 7.
num = 1

while num <= 10:
    print("7 x", num, "=", 7 * num)
    num += 1
```

```python
#Print the multiplication table of a number stored in a variable.
number =int(input("enter a number:"))
num = 1

while num <= 10:
    print(number * num)
    num += 1
```

```python
#Find the factorial of a number using a while loop.
number = 5
i = 1
fact = 1

while i <= number:
    fact *= i
    i += 1

print("Factorial =", fact)

```

## 🔹 Counting & Conditions

Count how many numbers between 1 and 50 are divisible by 3.

Print numbers between 1 and 50 divisible by both 3 and 5.

Reverse a number (example: 1234 → 4321).

Count the number of digits in a number.

```python
#Count how many numbers between 1 and 50 are divisible by 3.
num = 1
count = 0

while num <= 50:
    if num % 3 == 0:
        print(num)
        count += 1
    num += 1

print("Total =", count)

        
```

```python
#Print numbers between 1 and 50 divisible by both 3 and 5.
num = 1
count = 0

while num <= 50:
    if num % 3 == 0:
        print(num)
        count += 1
    num += 1

print("Total =", count)
```

```python
#Reverse a number (example: 1234 → 4321).


num = 1234

reverse = 0

while num > 0:
    digit = num % 10        # get last digit
    reverse = reverse * 10 + digit
    num = num // 10         # remove last digit

print(reverse)
```

```python
#Count the number of digits in a number.
num = 12345
count = 0

while num > 0:
    num = num // 10   # remove last digit
    count += 1

print("Digits:", count)
```
