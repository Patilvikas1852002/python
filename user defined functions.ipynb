```python
#1.Create a function to check whether a number is prime.
def is_prime(n):
    if n <= 1:
        return False  # 0 and 1 are not prime
    for i in range(2, int(n**0.5) + 1):  # check divisors up to √n
        if n % i == 0:
            return False
    return True


num = 17
if is_prime(num):
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

```python
#2 Create a function to generate Fibonacci series up to n terms.
def fibonacci_series(n):
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series

# Example usage
n_terms = 10
print("Fibonacci series up to", n_terms, "terms:", fibonacci_series(n_terms))
```

```python
#3 Create a function to find largest element in a list.
def largest_in_list(lst):
    if not lst:
        return None  # Handle empty list
    max_value = lst[0]  # Assume first element is largest
    for num in lst:
        if num > max_value:
            max_value = num
    return max_value


numbers = [10, 45, 2, 99, 23]
print("Largest element in the list is:", largest_in_list(numbers))
```

```python
#4 Create a function to remove duplicates from a list.
def remove_duplicates(lst):
    unique_list = []
    for item in lst:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list


numbers = [1, 2, 2, 3, 4, 4, 5]
print("List after removing duplicates:", remove_duplicates(numbers))
```

```python
#5 Create a function to count frequency of each element in a list.
def element_frequency(lst):
    freq = {}
    for item in lst:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1
    return freq


numbers = [1, 2, 2, 3, 3, 3, 4]
print("Frequency of elements:", element_frequency(numbers))
```

```python
#7 Create a function to check whether a string is palindrome.
def is_palindrome(s: str) -> bool:
    # Normalize the string: remove spaces and make lowercase
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]

# Example usage:
print(is_palindrome("madam"))          # True
print(is_palindrome("racecar"))        # True
print(is_palindrome("hello"))          # False
print(is_palindrome("A man a plan a canal Panama"))  # True
```

```python
#8 Create a function to calculate simple interest.
def simple_interest(principal: float, rate: float, time: float) -> float:
    """
    Calculate Simple Interest.
    
    Formula: SI = (P * R * T) / 100
    where:
    P = Principal amount
    R = Annual interest rate (in %)
    T = Time (in years)
    """
    return (principal * rate * time) / 100

# Example usage:
print(simple_interest(1000, 5, 2))   #
```

```python
#9 Create a function to find second largest number in a list.
def second_largest(numbers: list):
    """
    Find the second largest number in a list.
    Returns None if there is no second largest.
    """
    if len(numbers) < 2:
        return None
    
    unique_numbers = list(set(numbers))
    
    if len(unique_numbers) < 2:
        return None
    
    unique_numbers.sort(reverse=True)
    return unique_numbers[1]

# Example usage:
print(second_largest([10, 20, 4, 45, 99]))   # Output: 45
print(second_largest([5, 5, 5, 5]))          # Output: None
print(second_largest([1]))                   # Output: None


   
```

```python
#10 Create a function that returns both quotient and remainder.
def divide_numbers(dividend: int, divisor: int) -> tuple:
    """
    Returns quotient and remainder of division.
    """
    if divisor == 0:
        raise ValueError("Divisor cannot be zero.")
    
    quotient = dividend // divisor   # Integer division
    remainder = dividend % divisor   # Modulus operator
    return quotient, remainder

# Example usage:
print(divide_numbers(10, 3))   # Output: (3, 1)
print(divide_numbers(25, 4))   # Output: (6, 1)
print(divide_numbers(100, 10)) # Output: (10, 0)
```

```python
#1 Create a recursive function to calculate factorial.
def factorial(n: int) -> int:
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

```python
#2 Create a recursive function for Fibonacci series.
def fibonacci(n: int) -> int:
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(6))  # Output: 8
```

```python
#3 Create a function to implement calculator using multiple operations.
def calculator(a: float, b: float, operation: str) -> float:
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        if b == 0:
            raise ValueError("Division by zero not allowed")
        return a / b
    else:
        raise ValueError("Invalid operation")

print(calculator(10, 5, "add"))       # Output: 15
print(calculator(10, 5, "divide"))    # Output: 2.0
```

```python
#4 Create a function that accepts list and returns even and odd numbers separately.
def separate_even_odd(numbers: list) -> tuple:
    evens = [n for n in numbers if n % 2 == 0]
    odds = [n for n in numbers if n % 2 != 0]
    return evens, odds

print(separate_even_odd([1, 2, 3, 4, 5, 6]))
# Output: ([2, 4, 6], [1, 3, 5])
```

```python
#5 Create a function to perform matrix addition.
def matrix_addition(A: list, B: list) -> list:
    return [[A[i][j] + B[i][j] for j in range(len(A[0]))] for i in range(len(A))]

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print(matrix_addition(A, B))  # Output: [[6, 8], [10, 12]]
```

```python
#6 Create a function to perform matrix multiplication.
def matrix_multiplication(A: list, B: list) -> list:
    result = [[0] * len(B[0]) for _ in range(len(A))]
    for i in range(len(A)):
        for j in range(len(B[0])):
            for k in range(len(B)):
                result[i][j] += A[i][k] * B[k][j]
    return result

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print(matrix_multiplication(A, B))  # Output: [[19, 22], [43, 50]]
```

```python
#7 Create a function decorator example.
import time

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before function execution")
        result = func(*args, **kwargs)
        print("After function execution")
        return result
    return wrapper

@my_decorator
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
```

```python
#8 Create a function to validate email format.
import re

def validate_email(email: str) -> bool:
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    return re.match(pattern, email) is not None

print(validate_email("test@example.com"))  # True
print(validate_email("invalid-email"))     # False
```

```python
#9 Create a function to flatten nested list.
def flatten_list(nested_list: list) -> list:
    flat = []
    for item in nested_list:
        if isinstance(item, list):
            flat.extend(flatten_list(item))
        else:
            flat.append(item)
    return flat

print(flatten_list([1, [2, [3, 4], 5], 6]))
# Output: [1, 2, 3, 4, 5, 6]
```

```python
#10 Create a function that logs function execution time
import time

def log_execution_time(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution time: {end - start:.6f} seconds")
        return result
    return wrapper

@log_execution_time
def sample_function(n):
    total = 0
    for i in range(n):
        total += i
    return total

print(sample_function(1000000))
```
