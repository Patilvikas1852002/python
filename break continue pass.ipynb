```python
#5 Print multiplication table of 5 but stop when result exceeds 25
num = 5
i = 1

while True:
    result = num * i
    
    if result > 25:
        break
    
    print(num, "x", i, "=", result)
    i += 1
```

## INTERMEDIATE LEVEL (10 Questions)


```python
#1 Print prime numbers between 1 and 50 using break inside nested loop.
for num in range(2, 51):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num)
```

```python
#2 Create a password checker that allows only 3 attempts (use break).
correct_password = "python123"

for attempt in range(3):
    password = input("Enter password: ")
    
    if password == correct_password:
        print("Access Granted")
        break
    else:
        print("Wrong password")
```

```python
#3 Print numbers from 1 to 100 but skip all multiples of 5.
for i in range(1,101):
    if i%5==0:
        continue
    print(i)
```

```python
#4 Search for a number in a list and stop when found (use break).
numbers = [10, 25, 30, 45, 50]
search = int(input("Enter number to search: "))

for num in numbers:
    if num == search:
        print("Number found!")
        break
else:
    print("Number not found.")

```

```python
#5 Print Fibonacci series but stop if value exceeds 100.

a = 0
b = 1

while True:
    if a > 100:
        break
    
    print(a)
    
    c = a + b
    a = b
    b = c
```

```python
#6 Print all characters of a string except vowels using continue.
text = input("Enter a string: ")

for ch in text:
    if ch.lower() in "aeiou":
        continue
    print(ch)
```

```python
while True:
    print("\nMenu:")
    print("1. Say Hello")
    print("2. Say Goodbye")
    print("0. Exit")
    
    choice = input("Enter your choice: ")
    
    if choice == "0":
        print("Exiting program...")
        break
    elif choice == "1":
        print("Hello!")
    elif choice == "2":
        print("Goodbye!")
    else:
        print("Invalid choice. Try again.")
```

```python
#9 Loop through a list and use pass for negative numbers.
numbers = [5, -3, 7, -1, 0, 8]

for num in numbers:
    if num < 0:
        pass  # do nothing for negative numbers
    else:
        print(num)
```

```python
#10 Print star pattern but skip printing when row number is 3
rows = 5

for i in range(1, rows + 1):
    if i == 3:
        continue  # skip row 3
    print("* " * i)
```

## ADVANCED LEVEL (10 Questions)


```python
#1 Implement a number guessing game using break when correct guess is made.
secret = 7

while True:
    guess = int(input("Guess the number: "))
    if guess == secret:
        print("Correct! You win!")
        break
    else:
        print("Try again.")

```

```python
#2 Validate user input in a loop until correct format is entered.
while True:
    user_input = input("Enter a positive integer: ")
    if user_input.isdigit() and int(user_input) > 0:
        print("Valid input:", user_input)
        break
    else:
        print("Invalid input. Try again.")

```

```python
#3 Create ATM simulation with 3 password attempts using break.
correct_pin = "1234"

for attempt in range(3):
    pin = input("Enter ATM PIN: ")
    if pin == correct_pin:
        print("Access granted.")
        break
    else:
        print("Wrong PIN.")
else:
    print("ATM card blocked.")

```

```python
#4 Use nested loops and break outer loop when condition is satisfied.
for i in range(1, 6):
    for j in range(1, 6):
        print(i, j)
        if i * j > 10:
            print("Condition met, breaking both loops")
            break  # breaks inner loop
    else:
        continue  # continue outer loop if inner loop did not break
    break  # breaks outer loop if inner loop broke
```

```python
#5 Generate pairs from list but skip pairs where elements are equal.
numbers = [1, 2, 3]

for x in numbers:
    for y in numbers:
        if x == y:
            continue
        print(f"Pair: ({x}, {y})")
```

```python
#6 Process student marks list and skip absent students using continue.
marks = [45, None, 78, None, 90]  # None = absent

for mark in marks:
    if mark is None:
        continue
    print("Mark:", mark)
```

```python
#7 Implement infinite loop and safely exit using break.
while True:
    command = input("Type 'exit' to stop: ")
    if command.lower() == "exit":
        print("Exiting loop safely.")
        break
    print("You typed:", command)
```

```python
#8 Create login system that blocks after 3 wrong attempts.
correct_password = "secure123"

for attempt in range(3):
    pw = input("Enter password: ")
    if pw == correct_password:
        print("Login successful!")
        break
    else:
        print("Wrong password.")
else:
    print("Account blocked after 3 failed attempts.")

```

```python
#9 Implement search in matrix and break both loops once found.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
target = 5
found = False

for row in matrix:
    for val in row:
        if val == target:
            print("Found:", val)
            found = True
            break
    if found:
        break
```
