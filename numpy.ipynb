```python
#array reshaping and flattening
```

```python
arr1=np.array([1,2,3,4,5,6])
print(arr1)
```

```python
reshaped=arr1.reshape((6,1))   #shape=1,5(1 rows & 5 columns) dimension=1
                               #shape=5,1(5 rows & 1 columns)  dimension=2
print(reshaped)
```

```python
reshaped2=arr1.reshape((2,3))  #reshaping 2 dimensional array
print(reshaped2)
```

```python
#falatnning #it changes multilevel or higher dimension into one dimenssion array 
print(reshaped2.flatten())
```

```python
#array stacking and splitting
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

print(np.vstack((a,b))) #vertical stacking row-wise
print(np.hstack((a,b)))  #horizontal stacking column-wise
```

```python
#splitting
c=np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
split=np.hsplit(c,3)  #horizontal split 
for s in split:
    print(s)
```

```python
split=np.vsplit(c,2)  #vertical split 
for s in split:
    print(s)
```

```python
b=np.array([1,4,9])
print(b)
```

```python
#mathematical operations on  multiple array
```

```python
print(np.dot(a,b)) #3*1=3 + 2*4=8 + 3*5=15    =3+8+15=26   #dot product
```
